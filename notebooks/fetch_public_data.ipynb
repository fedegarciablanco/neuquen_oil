# Recursos públicos: producción de petróleo y gas por pozo

Este notebook consulta el catálogo de [Datos Argentina](https://datos.gob.ar/) para obtener los recursos publicados del dataset **Producción de petróleo y gas por pozo (Capítulo IV)**.

La consulta solo descarga los metadatos del catálogo. Las URLs de los archivos se muestran en una tabla para elegir qué recurso cargar después.

```python
from pathlib import Path
import requests

DATA_DIR = Path("../data")
DATA_DIR.mkdir(exist_ok=True)

FILES = {
    "pozos": DATA_DIR / "pozos.csv",
    "operadores": DATA_DIR / "operadores.csv",
    "pozos_shp": DATA_DIR / "pozos.zip",
}
```

```python
def download_file(url, destination, timeout=120):
    destination = Path(destination)

    print(f"Descargando:")
    print(f"  {url}")
    print(f"→ {destination}")

    response = requests.get(
        url,
        timeout=timeout,
        stream=True,
    )

    response.raise_for_status()

    with open(destination, "wb") as f:
        for chunk in response.iter_content(chunk_size=1024 * 1024):
            if chunk:
                f.write(chunk)

    print(f"OK: {destination.stat().st_size:,} bytes")
```
