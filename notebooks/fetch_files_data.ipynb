```python
import sys, os

repo_root = os.getcwd()
if repo_root not in sys.path:
    sys.path.insert(0, repo_root)

from utils.kml_to_xls import kml_or_kmz_to_dataframes, dataframes_to_excel

# --- Carpetas de salida ---
carpeta_ypf = os.path.join(repo_root, "generated_files", "ypf")
carpeta_pae = os.path.join(repo_root, "generated_files", "pae")
os.makedirs(carpeta_ypf, exist_ok=True)
os.makedirs(carpeta_pae, exist_ok=True)

# --- YPF ---
capas_ypf = kml_or_kmz_to_dataframes(
    os.path.join(repo_root, "source_files", "ypf", "diagrama_con_pozos_ypf.kml"),
    agrupar_por_carpeta=True
)
for nombre, df in capas_ypf.items():
    print(nombre, len(df), "filas")

dataframes_to_excel(capas_ypf, os.path.join(carpeta_ypf, "ypf_tabla.xlsx"))

# --- PAE ---
capas_pae = kml_or_kmz_to_dataframes(
    os.path.join(repo_root, "source_files", "pae", "pozos_pae.kmz"),
    agrupar_por_carpeta=False
)
for nombre, df in capas_pae.items():
    print(nombre, len(df), "filas")

dataframes_to_excel(capas_pae, os.path.join(carpeta_pae, "pae_tabla.xlsx"))
```
